## 📚 Import Libraries

## 📂 Load Dataset

The Superstore Sales dataset is loaded into a pandas DataFrame using the `read_csv()` function. This allows us to access, analyze, and manipulate the data efficiently throughout the project.


```python
df = pd.read_csv("Superstore.csv",encoding="latin1")
```

## 👀 Display the First 5 Rows

The `head()` function displays the first five rows of the dataset. This helps verify that the dataset has been loaded correctly and provides a quick overview of the data structure.


```python
df.head()
```

## 👀 Display the Last 5 Rows

The `tail()` function displays the last five rows of the dataset. This helps verify that the dataset has been loaded completely.

```python
df.tail()
```

## 📏 Check Dataset Shape

The `shape` attribute returns the number of rows and columns in the dataset. It helps us understand the size of the dataset before performing any analysis.


```python
df.shape
```

## 🏷️ Display Column Names

The `columns` attribute displays the names of all columns in the dataset. It helps us understand the available features for analysis.


```python
df.columns
```

## 📋 Dataset Information

The `info()` function provides a summary of the dataset, including the number of rows, column names, data types, non-null values, and memory usage. It is useful for understanding the structure and quality of the data.

```python
df.info()
```

## 📊 Statistical Summary

The `describe()` function generates descriptive statistics for numerical columns, such as count, mean, standard deviation, minimum, maximum, and quartile values. It provides a quick summary of the dataset.

```python
df.describe()
```

## 🔍 Check Missing Values

The `isnull().sum()` function checks for missing (null) values in each column of the dataset. Identifying missing values is an important step before data cleaning and analysis.

```python
df.isnull().sum()
```

## 🔄 Check Duplicate Rows

The `duplicated().sum()` function counts the number of duplicate rows in the dataset. Duplicate records should be identified before analysis to ensure accurate results.

```python
df.duplicated().sum()
```
