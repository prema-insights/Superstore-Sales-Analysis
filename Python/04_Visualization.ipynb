## 📚 Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 📂 Load Cleaned Dataset

Load the cleaned Superstore Sales dataset for visualization.

```python
df = pd.read_csv("Superstore_Cleaned.csv", encoding="latin1")

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

## 📊 Sales by Category

Create a bar chart to compare the total sales across different product categories.

```python
sales_by_category = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(8,5))
plt.bar(sales_by_category.index, sales_by_category.values)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

## 📈 Monthly Sales Trend

Create a line chart to visualize the monthly sales trend. This helps identify seasonal patterns and sales fluctuations over time.

```python
monthly_sales = df.groupby(df["Order Date"].dt.month)["Sales"].sum()

plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.show()
```

## 🥧 Sales Share by Category

Create a pie chart to visualize the percentage contribution of each product category to total sales.

```python
sales_by_category = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(7,7))
plt.pie(
    sales_by_category.values,
    labels=sales_by_category.index,
    autopct="%1.1f%%"
)

plt.title("Sales Share by Category")

plt.show()
```

## 📊 Sales Distribution

Create a histogram to visualize the distribution of sales values. This helps understand how sales are spread across different transactions.

```python
plt.figure(figsize=(8,5))

plt.hist(df["Sales"], bins=30)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

## 📦 Profit Distribution

Create a box plot to identify the distribution of profit values and detect potential outliers.

```python
plt.figure(figsize=(8,5))

plt.boxplot(df["Profit"])

plt.title("Profit Distribution")

plt.ylabel("Profit")

plt.show()
```

## 🏆 Top 10 Customers by Sales

Create a bar chart to visualize the top 10 customers based on total sales. This helps identify high-value customers for retention and loyalty programs.

```python
top_customers = (
    df.groupby("Customer Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))
plt.bar(top_customers.index, top_customers.values)

plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer Name")
plt.ylabel("Total Sales")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

## 📦 Top 10 Products by Sales

Visualize the top 10 products based on total sales to identify the best-selling products.

```python
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(11,5))
plt.bar(top_products.index, top_products.values)

plt.title("Top 10 Products by Sales")
plt.xlabel("Product Name")
plt.ylabel("Total Sales")

plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

## 🌍 Sales by Region

Compare the total sales across different regions using a bar chart.

```python
sales_region = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(7,5))
plt.bar(sales_region.index, sales_region.values)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

## 📈 Profit by Category

Compare the total profit across product categories to identify the most profitable category.

```python
profit_category = df.groupby("Category")["Profit"].sum()

plt.figure(figsize=(7,5))
plt.bar(profit_category.index, profit_category.values)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```
