## 📚 Import Libraries

## 📂 Load Cleaned Dataset

Load the cleaned Superstore Sales dataset for exploratory data analysis.

```python
df = pd.read_csv("Superstore_Cleaned.csv",encoding="latin1")
```

## 📋 Overall Dataset Summary

Before performing detailed analysis, it is important to understand the overall characteristics of the dataset. This section provides a quick summary of the dataset, including the number of records, columns, and basic business metrics.

```python
print("Total Rows :", df.shape[0])
print("Total Columns :", df.shape[1])
```

## 💰 Total Sales

Calculate the total sales generated by the Superstore.

```python
round(df["Sales"].sum(),2)
```

## 📈 Total Profit

Calculate the total profit earned by the Superstore.

```python
df["Profit"].sum()
```

## 👥 Total Customers

## 📦 Total Products

## 📊 Sales by Category

Analyze the total sales for each product category using the `groupby()` function.

```python
df.groupby("Category")["Sales"].sum()
```

## 🌍 Sales by Region

Calculate the total sales for each region to identify the best-performing regions.

```python
df.groupby("Region")["Sales"].sum()
```

## 👥 Sales by Customer Segment

Analyze the total sales across different customer segments.

```python
df.groupby("Segment")["Sales"].sum()
```

## 📊 Top Categories by Sales

Sort the categories based on total sales in descending order to identify the highest-performing category.

```python
round(df.groupby("Category")["Sales"].sum(),2).sort_values(ascending=False)
```

## 🌍 Sales by Region

Analyze the total sales generated in each region. This helps identify the regions that contribute the most to overall sales and supports regional business performance analysis.

```python
round(df.groupby("Region")["Sales"].sum(),2).sort_values(ascending=False)
```

## 👥 Sales by Customer Segment

Analyze the total sales across different customer segments. This helps understand which customer segment generates the highest sales for the business.

```python
df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
```

## 📊 Sales Statistics by Category

Calculate multiple sales statistics such as total sales, average sales, minimum sales, and maximum sales for each product category using the `agg()` function.

```python
round(df.groupby("Category")["Sales"].agg(["sum", "mean", "min", "max"]),2)
```

## 📊 Sales by Category and Region

Create a pivot table to analyze total sales for each product category across different regions. This provides a clear comparison of sales performance by category and region.

```python
round(pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum"
),2)
```

## 📈 Average Profit by Customer Segment

Create a pivot table to calculate the average profit for each customer segment.

```python
round(pd.pivot_table(
    df,
    values="Profit",
    index="Segment",
    aggfunc="mean"
),2)
```

## 🚚 Total Sales by Ship Mode

Create a pivot table to calculate the total sales for each shipping mode.

```python
pd.pivot_table(
    df,
    values="Sales",
    index="Ship Mode",
    aggfunc="sum"
)
```

## 🏆 Top 10 Customers by Sales

Identify the top 10 customers based on total sales. This helps the business recognize its highest-value customers and develop customer retention strategies.


```python
top_customers = (
    df.groupby("Customer Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

## 📦 Top 10 Products by Sales

Identify the top 10 products based on total sales. This helps understand which products generate the highest revenue.

```python
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

## 🌍 Top 10 States by Sales

Analyze the top 10 states based on total sales to identify the best-performing markets.

```python
top_states = (
    df.groupby("State")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_states
```

## 🏙️ Top 10 Cities by Profit

Identify the top 10 cities based on total profit to understand the most profitable locations.

```python
top_cities = (
    df.groupby("City")["Profit"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_cities
```

## 📅 Sales by Year

Analyze the total sales generated each year. This helps identify yearly sales trends and overall business growth.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

round(df.groupby(df["Order Date"].dt.year)["Sales"].sum(),2)
```

## 📅 Sales by Month

Analyze the total sales for each month to identify seasonal sales patterns.

```python
df.groupby(df["Order Date"].dt.month)["Sales"].sum()
```

## 📈 Profit by Year

Analyze the total profit earned each year to evaluate business profitability over time.

```python
df.groupby(df["Order Date"].dt.year)["Profit"].sum()
```

## 📦 Orders by Year

Count the total number of unique orders placed each year.

```python
df.groupby(df["Order Date"].dt.year)["Order ID"].nunique()
```
