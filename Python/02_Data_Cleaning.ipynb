```python
## 📚 Import Libraries
```

```python
import pandas as pd
import numpy as np
```

## 📂 Load Dataset

Load the Superstore Sales dataset into a pandas DataFrame for data cleaning.


```python
df = pd.read_csv("Superstore.csv",encoding="latin1")
```

## 🔍 Check Missing Values

Missing values can affect data analysis and visualization. Before cleaning the dataset, we need to identify columns that contain missing values.


```python
df.isnull().sum()
```

##Duplicate values


```python
df.duplicated().sum()
```

## 🗓️ Convert Date Columns

The `Order Date` and `Ship Date` columns are initially stored as text (`object`). Converting them to the `datetime` data type enables date-based analysis, filtering, and time series operations.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

## 🏷️ Verify Data Types

After converting the date columns, verify the data types to ensure that the conversion was successful.

```python
df.dtypes
```

## 💾 Save the Cleaned Dataset

After completing the data cleaning process, the cleaned dataset is saved as a new CSV file. This dataset will be used in the next stage for Exploratory Data Analysis (EDA).

```python
df.to_csv("Superstore_Cleaned.csv", index=False)
```
